```python
import random

numbers = [random.randint(1, 100) for _ in range(10)]

print("List of random numbers:")
print(numbers)

print("Even numbers:")
for number in numbers:
    if number % 2 == 0:
        print(number)
```

```python
import numpy as np

# Generates 5 random floating-point numbers between 0.0 and 1.0
random_floats = np.random.rand(5)
print("Random Floats:")
print(random_floats)

# Alternatively, generates 5 random integers between 1 and 100
random_ints = np.random.randint(1, 101, size=5)
print("\nRandom Integers:")
print(random_ints)
# Create an array of 5 random integers between 1 and 100
data = np.random.randint(1, 101, size=5)
print("Dataset:", data)

# 1. Mean
mean_val = np.mean(data)       # or data.mean()

# 2. Population Standard Deviation (default: ddof=0)
std_val = np.std(data)          # or data.std()

# Sample Standard Deviation (Bessel's correction: ddof=1)
sample_std_val = np.std(data, ddof=1)

print(f"Mean: {mean_val:.2f}")
print(f"Population Std Dev (N):   {std_val:.2f}")
print(f"Sample Std Dev (N - 1):   {sample_std_val:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create 500 equally spaced time values from 0 to 10
time = np.linspace(0, 10, 500)

# Calculate the amplitude of the sine wave
amplitude = np.sin(2 * np.pi * time)

# Plot the sine wave
plt.plot(time, amplitude)

# Add labels to the graph
plt.xlabel("Time")
plt.ylabel("Amplitude")

# Add title
plt.title("Sine Wave")

# Display the graph
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate the engineering data
# 500 points from 0 to 10 seconds
time = np.linspace(0, 10, 500)

# Damped sine wave
amplitude = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)

# 2. Create the Canvas (Figure) and Coordinate Box (Axes) explicitly
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

# 3. Plot the data with customized engineering styles
ax.plot(
    time,
    amplitude,
    color='crimson',
    linewidth=2,
    label='Sensor A Displacement'
)

# 4. Professional Engineering Customizations
ax.set_title(
    'Damped Harmonic Response of Structure',
    fontsize=14,
    fontweight='bold',
    pad=15
)

ax.set_xlabel(
    'Time (seconds)',
    fontsize=12
)

ax.set_ylabel(
    'Displacement (mm)',
    fontsize=12
)

# Enable engineering grid lines
ax.grid(
    True,
    linestyle='--',
    alpha=0.6
)

# Configure the Legend
ax.legend(
    loc='upper right',
    frameon=True,
    shadow=True
)

# 5. Improve layout
plt.tight_layout()

# 6. Display the graph
plt.show()
```

```python
ax.plot(time, amplitude, color='crimson', linewidth=2,
        label="Sensor A Displacement")

ax.set_title('Damped Harmonic Response of Structure',
             fontsize=14, fontweight='bold', pad=15)

ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_ylabel('Displacement (mm)', fontsize=12)

# Enable engineering gridlines
ax.grid(True, linestyle='--', alpha=0.6)

# Configure the Legend
ax.legend(loc='upper right', frameon=True, shadow=True)
```
