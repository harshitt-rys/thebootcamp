```python
import random

numbers = [random.randint(1, 100) for _ in range(10)]

print("List of random numbers:")
print(numbers)

print("Even numbers:")
for number in numbers:
    if number % 2 == 0:
        print(number)
```

```python
text = "Python is very easy"

print("Original string:")
print(text)

print("\nString in reverse:")
print(text[::-1])

print("\nWords in reverse order:")
words = text.split()
print(" ".join(words[::-1]))
```
