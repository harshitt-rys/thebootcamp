```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("Smart Transit Hub at Dusk(1).png")

if img is None:
    raise FileNotFoundError("Image file not found")

print("Image type:", type(img))
print("Image shape:", img.shape)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 7))
plt.imshow(img_rgb)
plt.axis("off")
plt.title("AI Generated Smart Transit Hub")
plt.show()
```

```python
%pip install -q ultralytics opencv-python

import cv2
from ultralytics import settings, YOLO

settings.update({"sync": False})

model = YOLO("yolov8n.pt")

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError("Could not open the webcam. Check camera permission.")

print("Starting Object Counter... Press 'q' to quit.")

while cap.isOpened():
    success, frame = cap.read()

    if not success:
        print("Failed to grab camera frame.")
        break

    results = model(frame, verbose=False)

    detected_count = len(results[0].boxes)

    annotated_frame = results[0].plot()

    cv2.putText(
        annotated_frame,
        f"Total Objects Counted: {detected_count}",
        (20, 50),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2,
        cv2.LINE_AA
    )

    cv2.imshow(
        "BCA AI Demo - Live Object Counter",
        annotated_frame
    )

    if cv2.waitKey(1) & 0xFF == ord("q"):
        print("Terminating program...")
        break

cap.release()
cv2.destroyAllWindows()

print("Camera feed closed successfully.")
```

```python
%pip install -q wordcloud

from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Words related to Artificial Intelligence
text = """
Artificial Intelligence Machine Learning Deep Learning
Computer Science Data Analytics Neural Networks
Natural Language Processing NLP Python Programming
Robotics Automation Computer Vision Speech Recognition
Prediction Classification Clustering Algorithms
Training Testing Model Knowledge Technology
Innovation Big Data Artificial Intelligence
AI Machine Learning Machine Learning
Data Data Data Computer Computer Learning Learning
Python Python Neural Network Network
"""

# Create the word cloud
wordcloud = WordCloud(
    width=1400,
    height=900,
    background_color="white",
    colormap="viridis",
    max_words=100,
    min_font_size=12,
    max_font_size=180,
    random_state=42
).generate(text)

# Display it
plt.figure(figsize=(14, 9))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")

plt.title(
    "Artificial Intelligence Word Cloud",
    fontsize=24,
    pad=20
)

plt.tight_layout()

# Save the image
plt.savefig(
    "AI_Word_Cloud.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
